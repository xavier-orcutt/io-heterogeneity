# Calculating RMST Pseudo-Observations

**This notebook calculates RMST pseudo-observations at 1 and 2 years for advanced head and neck cancer patients receiving pembrolizumab plus chemotherapy or pembrolizumab.**

```python
import sys
sys.path.append('../..')

import pandas as pd

from utils.pseudo_obs import pseudo_observations_rmst
```

```python
df = pd.read_csv('../outputs/pembrochemo_pembro_features_df.csv')
```

```python
df = df.set_index('PatientID')
```

```python
df.shape
```

```python
treatment_df = pd.read_csv('../outputs/pembrochemo_pembro_index.csv')
```

```python
treatment_df.shape
```

```python
df = pd.merge(df, treatment_df, on = 'PatientID', how = 'left')
```

```python
df.shape
```

```python
df['StartDate'] = pd.to_datetime(df['StartDate'])
```

```python
df['treatment_year'] = df['StartDate'].dt.year
```

```python
df = df.query('treatment_year <= 2022')
```

```python
df.shape
```

```python
pseudo_rmst = pseudo_observations_rmst(
    df['duration'].values,
    df['event'].values,
    tau = 365
)

df['rmst_pseudo_1y'] = pseudo_rmst
```

```python
pseudo_rmst = pseudo_observations_rmst(
    df['duration'].values,
    df['event'].values,
    tau = 730
)

df['rmst_pseudo_2y'] = pseudo_rmst
```

```python
df = df.reset_index()
```

```python
df = df[['PatientID', 'rmst_pseudo_1y', 'rmst_pseudo_2y']]
```

```python
df.head(5)
```

```python
df.to_csv('../outputs/pseudo_obs_rmst.csv', index = False)
```
